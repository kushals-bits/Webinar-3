```python
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

REPO_ROOT = Path("..").resolve()
sys.path.insert(0, str(REPO_ROOT))

from src.text_pipeline.pipeline import run_text_pipeline
from src.image_pipeline.pipeline import run_image_pipeline
from src.image_pipeline.signal_processor import run_signal_pipeline
from src.timeseries_pipeline.pipeline import run_timeseries_pipeline
from src.graph_pipeline.pipeline import run_graph_pipeline
from src.evaluation.comparison import build_master_comparison_report, print_master_comparison

print("[Setup] All multimodal pipeline subsystems loaded!")
```

## 1. Running the Complete 4-Track Multimodal Pipeline

```python
data_dir = REPO_ROOT / "data"

# 1. Module 9: Text
text_res = run_text_pipeline(str(data_dir / "text" / "patient_reports_raw.csv"))

# 2. Module 10: Image & Signal
img_res = run_image_pipeline(str(data_dir / "image" / "raw"))
sig_res = run_signal_pipeline(str(data_dir / "signal"))

# 3. Module 11: Time-Series
ts_res = run_timeseries_pipeline(str(data_dir / "timeseries" / "vitals_stream.csv"))

# 4. Module 12: Knowledge Graph
graph_res = run_graph_pipeline(str(data_dir / "graph"))
```

## 2. Master Multimodal Comparison Benchmark

```python
all_results = [text_res, img_res, sig_res, ts_res, graph_res]
master_df = build_master_comparison_report(all_results)
print_master_comparison(master_df)
display(master_df)
```

## 3. Visualizing Before vs After Performance Lift Across All 4 Modalities

```python
fig, ax = plt.subplots(figsize=(12, 6))

tracks = [m.split("(")[0].strip() for m in master_df["Curriculum Track"]]
x = np.arange(len(tracks))
w = 0.35

rects1 = ax.bar(x - w/2, master_df["Baseline F1-Macro"], w, label="Before (Raw / Naive)", color="#e74c3c", alpha=0.9)
rects2 = ax.bar(x + w/2, master_df["Preprocessed F1-Macro"], w, label="After (Sanitized & Engineered)", color="#2ecc71", alpha=0.9)

ax.set_xticks(x)
ax.set_xticklabels(tracks, fontweight="bold", fontsize=11)
ax.set_ylabel("F1-Macro Score", fontsize=12)
ax.set_title("HealthAI 360: Impact of Rigorous Preprocessing Across Modalities (Modules 9 - 12)", fontweight="bold", fontsize=14)
ax.set_ylim(0, 1.1)
ax.legend(frameon=True, facecolor="white", edgecolor="#ccc", fontsize=11)
ax.grid(axis="y", linestyle="--", alpha=0.4)

# Value annotations
for i in range(len(tracks)):
    b_val = master_df["Baseline F1-Macro"].iloc[i]
    a_val = master_df["Preprocessed F1-Macro"].iloc[i]
    lift = master_df["Delta F1 (Lift)"].iloc[i]
    ax.text(x[i] - w/2, b_val + 0.02, f"{b_val:.2f}", ha="center", fontsize=9)
    ax.text(x[i] + w/2, a_val + 0.02, f"{a_val:.2f}\n(+{lift:.2f})", ha="center", fontweight="bold", fontsize=9, color="#27ae60")

plt.tight_layout()
plt.show()
```
